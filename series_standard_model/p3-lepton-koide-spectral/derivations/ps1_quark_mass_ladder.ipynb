## Section 3: Testing Mechanisms A–E

```python
# Mechanism A: naive φ^3(l-1) scaling
quarks = ['strange','charm','bottom','top']
pdg = [93.4, 1270, 4180, 172760]
cage_n = [1, 2, 3, 4]
cpp_N  = [4, 12, 20, 60]

rows = []
cumul = 5.0
for name, n, N, p in zip(quarks, cage_n, cpp_N, pdg):
    E = N * E_eDP * phi**(3*(n-1))
    cumul += E
    rows.append({'Quark': name, 'N': N, 'φ^3(l-1)': round(phi**(3*(n-1)),2),
                 'M_struct (MeV)': round(cumul,1), 'PDG (MeV)': p,
                 'Ratio': round(cumul/p,2)})

print('Mechanism A: N × E_eDP × φ^3(l-1)')
print(pd.DataFrame(rows).to_string(index=False))
print('\n→ Errors 2–10×. FALSIFIED by exact shell computation.')
```

```python
# Mechanism B: PSR compression
print('Mechanism B: PSR compression r_eff = l_P/(1 + α·n·sea)')
for n in range(1,5):
    r = 1.0 / (1 + alpha_geom * n * sea_strength)
    print(f'  n={n}: r_eff/r = {r:.4f},  (r_eff/r)³ = {r**3:.4f}')
print('\n→ Factors 0.75–0.37. Too small to bridge gap.')

# Mechanism C: phase cancellation
print('\nMechanism C: C_n = |Σ exp(i·2πm/φ²)| / n')
for n in range(1,5):
    C = abs(sum(np.exp(1j*2*np.pi*m/phi**2) for m in range(n+1))) / n
    print(f'  n={n}: C_n = {C:.4f}')
print('\n→ Oscillates 0.72, 0.24, 0.36, 0.07 — quasi-random, not systematic.')
```

```python
# Mechanism E: combined A × B × C (Grok's proposal)
vol_3d = [0.515028, 1.809017, 2.181695, 3.266125]  # actual shell volumes

rows = []
cumul = 5.0
for name, n, N, p in zip(quarks, cage_n, cpp_N, pdg):
    naive = N * E_eDP * phi**(3*(n-1))
    psr   = (1.0 / (1 + alpha_geom * n * sea_strength))**3
    C_n   = abs(sum(np.exp(1j*2*np.pi*m/phi**2) for m in range(n+1))) / n
    combined = naive * psr * C_n
    cumul += combined
    rows.append({'Quark': name, 'Naive': round(naive,1), '×PSR³': round(naive*psr,1),
                 '×C_n': round(naive*psr*C_n,1), 'Cumul': round(cumul,1),
                 'PDG': p, 'Ratio': round(cumul/p,2)})

print('Mechanism E: combined naive × PSR³ × C_n')
print(pd.DataFrame(rows).to_string(index=False))
print('\n→ Still off by 0.07–2.0×. The "corrections" are larger than base terms.')
print('   Decomposition itself is incorrect — not a fixable numerical issue.')
```

Despite the negative result, two precise relations emerge.

```python
# Discovery 2: K(c,b,t) ≈ 2/3
def koide_K(masses):
    return sum(masses) / sum(np.sqrt(m) for m in masses)**2

triplets = [
    ('u, c, t', [2.16, 1270, 172760]),
    ('d, s, b', [4.70, 93.4, 4180]),
    ('c, b, t', [1270, 4180, 172760]),
    ('e, μ, τ', [0.511, 105.658, 1776.86]),
]

rows = []
for name, masses in triplets:
    K = koide_K(masses)
    dev = abs(K - 2/3) / (2/3) * 100
    rows.append({'Triplet': name, 'K': round(K,4), 'Dev from 2/3': f'{dev:.1f}%'})

print('Discovery 2: Koide K for various triplets')
print(pd.DataFrame(rows).to_string(index=False))
print()
print('K(c,b,t) is 25× closer to 2/3 than K(d,s,b).')
print('Physical interpretation: for heavy quarks, current mass >> cage binding,')
print('so the K3 spectral signature shows through. Consistent with Thomas\'s')
print('DP-cloud picture: mass is primarily ZBW thermal energy (K3 structure),')
print('with cage binding as a perturbation — large for light, small for heavy.')
```

## Section 5: φ-Exponent Analysis

```python
# Express all quark masses as m_e × φ^n
rows = []
for name, mass in pdg_current.items():
    phi_exp = np.log(mass / m_e) / np.log(phi)
    n_int   = round(phi_exp)
    pred    = m_e * phi**n_int
    err     = abs(pred - mass) / mass * 100
    rows.append({'Quark': name, 'PDG (MeV)': mass,
                 'φ exponent': round(phi_exp,2), 'n (nearest int)': n_int,
                 'm_e × φⁿ': round(pred,2), 'Error': f'{err:.1f}%'})

print('Quark masses expressed as m_e × φⁿ')
print(pd.DataFrame(rows).to_string(index=False))
print()
print('Exponent sequence: 3, 5, 11, 16, 19, 26')
print('Differences: 2, 6, 5, 3, 7 — no obvious arithmetic pattern')
print('Errors: systematically 10–20% (except u at 0.2%)')
print('Note: n=11 appears for BOTH m_s/m_e AND m_μ/m_e — suggestive')
```
